## Cell 4 — Download engine

````python
import os
import glob
import json
import shutil
import zipfile
import subprocess
import pandas as pd
from pathlib import Path

BASE_DIR = Path("datasets")
STAGING  = Path("/tmp/kaggle_staging")

for cat in ["tall", "wide_tall", "small", "wide"]:
    (BASE_DIR / cat).mkdir(parents=True, exist_ok=True)
STAGING.mkdir(parents=True, exist_ok=True)

ok, skipped, failed = [], [], []

def find_csv(folder, hint):
    if hint:
        exact = list(folder.rglob(hint))
        if exact:
            return exact[0]
    csvs = list(folder.rglob("*.csv"))
    if csvs:
        return sorted(csvs, key=lambda p: p.stat().st_size, reverse=True)[0]
    others = list(folder.rglob("*.data")) + list(folder.rglob("*.tsv"))
    return others[0] if others else None

def kaggle_download(entry):
    dest_path = BASE_DIR / entry["file"]


    stage = STAGING / entry["file"].replace("/", "_")
    stage.mkdir(parents=True, exist_ok=True)

    if entry["kaggle_type"] == "competition":
        cmd = ["kaggle", "competitions", "download",
               "-c", entry["kaggle_id"], "-p", str(stage)]
    else:
        cmd = ["kaggle", "datasets", "download",
               "-d", entry["kaggle_id"], "-p", str(stage)]

    print(f"  ⬇  {entry['file']}  ({entry['kaggle_id']})")
    result = subprocess.run(cmd, capture_output=True, text=True)

    if result.returncode != 0:
        err = result.stderr.strip() or result.stdout.strip()
        print(f"     ❌ download failed: {err[:120]}")
        failed.append({"file": entry["file"], "reason": err[:200]})
        return

    for z in stage.glob("*.zip"):
        with zipfile.ZipFile(z, 'r') as zf:
            zf.extractall(stage)
        z.unlink()

    src = find_csv(stage, entry.get("csv_name"))
    if src is None:
        msg = f"No CSV found in staging for {entry['file']}"
        print(f"     ❌ {msg}")
        failed.append({"file": entry["file"], "reason": msg})
        return

    # Read kwargs — handle separator, header injection, row limit
    row_limit   = entry.get("row_limit")
    sep         = entry.get("sep", ",")
    add_header  = entry.get("add_header")

    read_kwargs = {"sep": sep}
    if add_header:
        read_kwargs["header"] = None
        read_kwargs["names"] = add_header
    if row_limit:
        read_kwargs["nrows"] = row_limit

    needs_processing = bool(add_header or row_limit or sep != ",")

    if needs_processing:
        try:
            df = pd.read_csv(src, **read_kwargs)
            df.to_csv(dest_path, index=False)
            print(f"     ✅ saved {len(df)} rows × {len(df.columns)} cols → {dest_path}")
        except Exception as e:
            print(f"     ❌ post-process failed: {e}")
            failed.append({"file": entry["file"], "reason": str(e)})
            return
    else:
        shutil.copy2(src, dest_path)
        size_kb = dest_path.stat().st_size // 1024
        print(f"     ✅ saved {size_kb} KB → {dest_path}")

    ok.append(entry["file"])

print("Download engine ready.")
````

## Cell 5 — Run the downloads
This cell downloads all 40 datasets. It's safe to re-run — already-downloaded files are skipped automatically.

````python
ok.clear(); skipped.clear(); failed.clear()

for i, entry in enumerate(REGISTRY, 1):
    print(f"[{i:02d}/{len(REGISTRY)}]", end=" ")
    try:
        kaggle_download(entry)
    except Exception as e:
        print(f"     ❌ unexpected error: {e}")
        failed.append({"file": entry["file"], "reason": str(e)})

# ── Summary ──────────────────────────────────────────────────────────────
print("\n" + "="*60)
print(f"  ✅  Downloaded : {len(ok)}")
print(f"  ⏭  Skipped    : {len(skipped)}")
print(f"  ❌  Failed     : {len(failed)}")
if failed:
    print("\nFailed datasets:")
    for f in failed:
        print(f"  • {f['file']}")
        print(f"    reason: {f['reason'][:120]}")
print("="*60)
````

## Cell 6 — Validate downloads + fix target column names
Checks each CSV is readable, target column exists, and row/col counts meet the shape requirement.

````python
SHAPE_RULES = {
    "tall":      lambda r, c: r > 800,
    "wide_tall": lambda r, c: r > 800 and c > 14,
    "small":     lambda r, c: r < 200,
    "wide":      lambda r, c: r < 200  and c > 20,
}

validation_issues = []
config_entries    = []

print(f"{'File':<45} {'Shape':>12}  {'Target?':>8}  {'Shape OK?':>10}")
print("-" * 82)

for entry in REGISTRY:
    path = BASE_DIR / entry["file"]
    cat  = entry["file"].split("/")[0]

    if not path.exists():
        validation_issues.append((entry["file"], "file not found"))
        print(f"  {'❌ MISSING':<43} {'—':>12}  {'—':>8}  {'—':>10}")
        continue

    try:
        df = pd.read_csv(path, nrows=5)
        nrows = sum(1 for _ in open(path)) - 1   # fast line count
        ncols = len(df.columns)

        # Fuzzy target match (case-insensitive strip)
        col_map = {c.strip().lower(): c for c in df.columns}
        tgt_key  = entry["target"].strip().lower()
        tgt_found = tgt_key in col_map
        actual_col = col_map.get(tgt_key, entry["target"])

        shape_ok = SHAPE_RULES.get(cat, lambda r, c: True)(nrows, ncols)

        status = "✅" if (tgt_found and shape_ok) else "⚠️ "
        print(f"{status} {entry['file']:<43} {nrows:>6}×{ncols:<5}  "
              f"{'YES' if tgt_found else 'NO':>8}  "
              f"{'YES' if shape_ok else 'NO':>10}")

        issues = []
        if not tgt_found:
            issues.append(f"target '{entry['target']}' not in columns: {list(df.columns[:6])}")
        if not shape_ok:
            issues.append(f"shape {nrows}×{ncols} doesn't match '{cat}' rule")
        if issues:
            validation_issues.extend([(entry["file"], i) for i in issues])

        config_entries.append({"file": entry["file"], "target": actual_col})

    except Exception as e:
        validation_issues.append((entry["file"], str(e)))
        print(f"  ❌ READ ERROR  {entry['file']:<40}  {str(e)[:50]}")

print("-" * 82)
print(f"\nValidation issues: {len(validation_issues)}")
for f, reason in validation_issues:
    print(f"  • {f}: {reason}")
````

## Cell 7 — Write config.json
Only includes datasets that actually exist on disk.

````python
config_path = BASE_DIR / "config.json"
with open(config_path, "w") as f:
    json.dump(config_entries, f, indent=2)

print(f"✅ config.json written with {len(config_entries)} entries → {config_path}")
print()
print(json.dumps(config_entries[:5], indent=2), "\n  ...")
````

## Cell 8 — Retry failed datasets (run after Cell 5 if anything failed)
Re-runs only the entries that are still missing from disk.

````python
missing = [e for e in REGISTRY if not (BASE_DIR / e["file"]).exists()]

if not missing:
    print("🎉 All datasets already on disk — nothing to retry.")
else:
    print(f"Retrying {len(missing)} missing dataset(s)...\n")
    retry_ok, retry_fail = [], []
    for entry in missing:
        try:
            kaggle_download(entry)
            if (BASE_DIR / entry["file"]).exists():
                retry_ok.append(entry["file"])
            else:
                retry_fail.append(entry["file"])
        except Exception as e:
            print(f"  ❌ {entry['file']}: {e}")
            retry_fail.append(entry["file"])
    print(f"\nRetry result — recovered: {len(retry_ok)}, still missing: {len(retry_fail)}")
    if retry_fail:
        print("Still missing (manual download needed):")
        for f in retry_fail:
            e = next(x for x in REGISTRY if x["file"]==f)
            print(f"  • {f}  →  https://www.kaggle.com/datasets/{e['kaggle_id']}")
````
